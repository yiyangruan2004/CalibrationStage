```python
import pandas as pd
import matplotlib.pyplot as plt
sens = 2.149e-007 

def plot3D(csv_file):
    XYZ = pd.read_csv(csv_file)
    P = XYZ.Amp / 1e6 / sens
    fig = plt.figure(figsize=(14, 5))
    ax = fig.add_subplot(1, 2, 1, projection='3d')
    sc = ax.scatter(
        XYZ.X, XYZ.Y, XYZ.Z,
        c=P,
        cmap='turbo'
    )
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_zlabel('Z')
    ax.set_title("Coarse 3D Sound field")
    cbar = fig.colorbar(sc, ax=ax, pad=0.1)
    cbar.set_label('Pressure (kPa)')
    plt.tight_layout()
    plt.show()
plot3D("XYZ.csv")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
sens = 2.149e-7
def plot2D(csv_file, plane):
    planes = {
        "XY": ("X", "Y", "Z"),
        "XZ": ("X", "Z", "Y"),
        "YZ": ("Y", "Z", "X"),
    }
    cx, ix, slice_axis = planes[plane]
    data = pd.read_csv(csv_file)
    slice_val = data[slice_axis].median()
    data = data[np.isclose(data[slice_axis], slice_val)]
    data["P"] = data["Amp"] / 1e6 / sens

    # pivot to 2D grid
    p = data.pivot(index=ix, columns=cx, values="P")
    fig, ax = plt.subplots(figsize=(6, 5))
    im = ax.imshow(
        p.values,
        origin="lower",
        aspect="equal",
        extent=[p.columns.min(), p.columns.max(), p.index.min(), p.index.max()],
        cmap="turbo"
    )
    ax.set_title(f"{plane} plane")
    ax.set_xlabel(cx)
    ax.set_ylabel(ix)
    fig.colorbar(im, ax=ax, label="Pressure (kPa)")
    plt.tight_layout()
    plt.show()
plot2D("XYZ.csv", "XZ")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

def plot1D(csv_file, axis):
    data = pd.read_csv(csv_file)
    P = data["Amp"] / 1e6 / sens
    axes = ["X", "Y", "Z"]
    fig, ax = plt.subplots(figsize=(6, 4))

    x = data[axis]
    y = P
    df = pd.DataFrame({axis: x, "P": y})
    df = df.groupby(axis, as_index=False).mean()

    ax.plot(df[axis], df["P"])
    ax.set_xlabel(f"{axis} (inch)")
    ax.set_ylabel("Pressure (kPa)")
    ax.set_title(f"1D Pressure Profile along {axis}")

    plt.tight_layout()
    plt.show()

plot1D("XYZ.csv","Z")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data = pd.read_csv('Focus.csv')
p = data.iloc[:, 0].values / 1e6 / sens
dt = 0.016  # ns per sample
x = np.arange(len(p)) * dt  # time axis in ns
plt.plot(x, p)
plt.xlabel('Time (us)')
plt.ylabel('Pressure (kPa)')
plt.title('Representitive Signal')
plt.grid(True)
plt.show()
```
